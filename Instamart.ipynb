```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
dataset=pd.read_csv("Instamart_customers.csv")
dataset.head()
dataset.isnull().sum()
dataset['area']=dataset['area'].str.strip()
aov=dataset.groupby('area')['avg_order_value'].mean().round(2)
aov_sorted=aov.sort_values(ascending=False).head(20)
plt.figure(figsize=(20,10))

sns.barplot(
    x=aov_sorted.values,
    y=aov_sorted.index,
    hue=aov_sorted.index,
    palette='magma'
)
plt.title('Top 20 Regions by Average Order Value',fontsize=25)
plt.xlabel('Average Order Value-------->',fontsize=18)
plt.ylabel('Region ------->',fontsize=18)

for i, v in enumerate(aov_sorted.values):
    plt.text(v+5, i, f'₹{v:.0f}', va='center',fontsize=10)
plt.tight_layout()
plt.savefig('AOV_BY_REGION.png',dpi=300)
plt.show()
```
